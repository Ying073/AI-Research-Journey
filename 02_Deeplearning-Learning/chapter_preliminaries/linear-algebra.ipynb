# 线性代数

## 标量 由只有一个元素的张量表示

```python
import torch

x=torch.tensor([3.0])
y=torch.tensor([2.0])

x+y,x*y,x/y,x**y
```

## 可将向量视为标量组成的列表

```python
x=torch.arange(4)
x
```

## 通过张量的索引来访问任一元素

```python
x[3]
```

## 访问张量的长度

```python
len(x)
```

## 只有一个轴的张量，形状只有一个元素

```python
x.shape
```

## 通过指定两个分量m和n来创建一个形状为m x n的矩阵

```python
A=torch.arange(20).reshape(4,5)
A
```

## 矩阵的转置

```python
A.T
```

## 对称矩阵 A=其转置

```python
B=torch.tensor([[1,2,3],[2,0,4],[3,4,5]])
B
```

```python
B==B.T
```

## 向量是标量的推广，矩阵是向量的推广

```python
X=torch.arange(24).reshape(2,3,4)
X
```

## 给定具有相同形状的任何两个张量，任何按元素二元运算的结果都将是相同形状的张量

```python
A=torch.arange(20,dtype=torch.float32).reshape(5,4)
B=A.clone()  # 通过分配新的内存，将A的一个副本分配给B
A, A+B
```

## 两个矩阵的按元素乘法成为 哈达玛积（Hadamard Product） 

```python
A*B
```
